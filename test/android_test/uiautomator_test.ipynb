```python
from itertools import zip_longest

# 原始数据：每个子列表是一列（纵向）
columns = [
    [1, 2, 3],      # 第1列，长度3
    [4, 5],         # 第2列，长度2
    [6, 7, 8, 9]    # 第3列，长度4
]

# 转置成行：缺失处填 None（或自定义值）
rows = list(zip_longest(*columns, fillvalue=None))

print(rows)
# 输出：
# [(1, 4, 6),
#  (2, 5, 7),
#  (3, None, 8),
#  (None, None, 9)]
```
